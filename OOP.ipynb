```python
from abc import ABC, abstractmethod

# ==========================================
# LATIHAN 5: ABSTRACTION & INTERFACE
# ==========================================
# GameUnit bertindak sebagai Kontrak Utama (Abstract Class)
class GameUnit(ABC):
    @abstractmethod
    def serang(self, target):
        pass

    @abstractmethod
    def info(self):
        pass


# ==========================================
# LATIHAN 1, 2, & 4: CLASS HERO (ENKAPSULASI)
# ==========================================
class Hero(GameUnit):
    def __init__(self, nama, hp, attack_power):
        self.nama = nama
        self.attack_power = attack_power

        # Enkapsulasi: HP bersifat Private (Latihan 4)
        self.__hp = 0
        self.set_hp(hp)  # Menggunakan setter agar tervalidasi sejak awal

    # Getter untuk mengakses HP
    def get_hp(self):
        return self.__hp

    # Setter dengan validasi keamanan/integritas data
    def set_hp(self, nilai_baru):
        if nilai_baru < 0:
            self.__hp = 0  # HP tidak boleh negatif
        elif nilai_baru > 1000:
            print("Cheat terdeteksi! HP dimaksimalkan ke 1000 saja.")
            self.__hp = 1000
        else:
            self.__hp = nilai_baru

    # Implementasi Method Kontrak dari GameUnit
    def info(self):
        print(f"Hero: {self.nama} | HP: {self.get_hp()} | Power: {self.attack_power}")

    def serang(self, lawan):
        # Latihan 2: Menerima objek utuh pada parameter 'lawan'
        if isinstance(lawan, GameUnit):
            print(f"\n{self.nama} menyerang {lawan.nama if hasattr(lawan, 'nama') else lawan.jenis}!")
            lawan.diserang(self.attack_power)
        else:
            print(f"\n{self.nama} menyerang {lawan}!")

    def diserang(self, damage):
        sisa_hp = self.get_hp() - damage
        self.set_hp(sisa_hp)
        print(f"{self.nama} terkena damage {damage}. Sisa HP: {self.get_hp()}")


# Class Monster untuk membuktikan implementasi GameUnit (Latihan 5)
class Monster(GameUnit):
    def __init__(self, jenis, hp):
        self.jenis = jenis
        self.hp = hp

    def serang(self, target):
        print(f"Monster {self.jenis} menggigit {target}!")

    def info(self):
        print(f"Saya adalah Monster: {self.jenis} | HP: {self.hp}")


# ==========================================
# LATIHAN 3 & 6: PEWARISAN & POLIMORFISME
# ==========================================
# Child Class 1: Mage (Memilih warisan Hero & menambah fitur Mana/Skill)
class Mage(Hero):
    def __init__(self, nama, hp, attack_power, mana):
        # Menggunakan super() untuk menghubungkan data ke class Induk (Hero)
        super().__init__(nama, hp, attack_power)
        self.mana = mana

    # Override method info() untuk menampilkan Mana
    def info(self):
        print(f"{self.nama} [Mage] | HP: {self.get_hp()} | Power: {self.attack_power} | Mana: {self.mana}")

    # Skill khusus Mage
    def skill_fireball(self, lawan):
        if self.mana >= 20:
            print(f"\n{self.nama} menggunakan Fireball ke {lawan.nama}!")
            self.mana -= 20
            lawan.diserang(self.attack_power * 2)  # Damage 2x lipat
        else:
            print(f"\n{self.nama} gagal skill! Mana tidak cukup.")

    # Override method serang untuk Polimorfisme
    def serang(self, lawan=None):
        if lawan and isinstance(lawan, GameUnit):
            super().serang(lawan)
        else:
            print(f"{self.nama} (Mage) menembakkan Bola Api! Boom!")


# Child Class 2: Archer
class Archer(Hero):
    def serang(self, lawan=None):
        if lawan and isinstance(lawan, GameUnit):
            super().serang(lawan)
        else:
            print(f"{self.nama} (Archer) memanah dari jauh! Jleb!")


# Child Class 3: Fighter
class Fighter(Hero):
    def serang(self, lawan=None):
        if lawan and isinstance(lawan, GameUnit):
            super().serang(lawan)
        else:
            print(f"{self.nama} (Fighter) memukul dengan pedang! Slash!")


# Child Class 4: Healer (Tugas Analisis 6: Uji Skalabilitas)
class Healer(Hero):
    def serang(self, lawan=None):
        print(f"{self.nama} (Healer) tidak menyerang, tapi menyembuhkan teman!")


# ==========================================
# MAIN PROGRAM (PENGUJIAN SELURUH FITUR)
# ==========================================
if __name__ == "__main__":
    print("=== 1. UJI LATIHAN 1 & 2 (INSTANSIASI & SERANGAN) ===")
    hero1 = Hero("Layla", 100, 15)
    hero2 = Hero("Zilong", 120, 20)

    hero1.info()
    hero2.info()

    print("\n--- Pertarungan Dimulai ---")
    hero1.serang(hero2)  # Layla menyerang Zilong
    hero2.serang(hero1)  # Zilong membalas


    print("\n=== 2. UJI LATIHAN 3 (INHERITANCE & SUPER) ===")
    eudora = Mage("Eudora", 80, 30, 100)
    balmond = Hero("Balmond", 200, 10)

    eudora.info()
    eudora.serang(balmond)         # Serangan biasa
    eudora.skill_fireball(balmond) # Skill khusus Mage


    print("\n=== 3. UJI LATIHAN 4 (ENKAPSULASI & VALIDASI) ===")
    print("Mencoba set HP negatif (-50):")
    hero1.set_hp(-50)
    print(f"Hasil HP Hero1: {hero1.get_hp()}")  # Mencegat jadi 0

    print("Mencoba set HP Cheat (9999):")
    hero1.set_hp(9999)
    print(f"Hasil HP Hero1: {hero1.get_hp()}")  # Mencegat jadi 1000

    # Pembuktian Name Mangling (Tugas Analisis 4)
    print(f"Mencoba akses paksa Name Mangling: {hero1._Hero__hp}")


    print("\n=== 4. UJI LATIHAN 5 (ABSTRACTION & INTERFACE) ===")
    h_alucard = Hero("Alucard", 150, 25)
    m_serigala = Monster("Serigala", 80)

    h_alucard.info()
    m_serigala.info()


    print("\n=== 5. UJI LATIHAN 6 (POLYMorphism) ===")
    pasukan = [
        Mage("Eudora", 80, 30, 100),
        Archer("Miya", 90, 25),
        Fighter("Zilong", 120, 20),
        Healer("Rafaela", 70, 10)  # Menambahkan Healer (Hasil Analisis 6)
    ]

    print("--- PERANG DIMULAI ---")
    # Satu perintah loop, tetapi respon menyesuaikan tipe kelas masing-masing
    for pahlawan in pasukan:
        pahlawan.serang()
```
